```python
import numpy as np
from sklearn.metrics.pairwise import cosine_similarity

#                      spicy  sweet  crunchy
dishes = {
    "Gulab Jamun":    [0.05,  0.95,  0.05],
    "Rasmalai":       [0.10,  0.85,  0.00],
    "Butter Chicken": [0.60,  0.20,  0.10],
    "Paneer Tikka":   [0.70,  0.10,  0.30],
    "Samosa":         [0.50,  0.00,  0.90],
}

query = np.array([[0.0, 1.0, 0.0]])         # "something sweet": all weight on sweetness
print(query)
names = list(dishes)
matrix = np.array([dishes[n] for n in names])
scores = cosine_similarity(query, matrix)[0]

for name, score in zip(names, scores):
    print(f"{name} {score:.3f}")
```
