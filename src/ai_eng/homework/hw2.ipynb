```python
class AgentSession:
    def __init__(self, user_id):
        self.user_id = user_id

    def start(self):
        self.is_active = True
        print(f"Session started for user {self.user_id}.")

    def end(self):
        self.is_active = False
        print(f"Session ended for user {self.user_id}.")

    def __del__(self):
        """Destructor method called when the object's memory is garbage collected."""
        print(f"AgentSession object for user {self.user_id} has been destroyed.")

# 1. Create the instance with only user_id initially
session = AgentSession(user_id="usr_9042")
print("Initial attributes:", session.__dict__)

session.model = "gpt-4o-mini"
session.message_count = 0
session.is_active = False
print("Attributes after dynamic addition:", session.__dict__)

session.start()
print("status after starting session:", session.is_active)

session.end()
print("status after ending session:", session.is_active)

print("\nDeleting session object...")
del session
```

```python
class AIModel:
    def __init__(self, name, provider):
        self.name = name
        self.provider = provider

    @classmethod
    def from_string(cls, model_string: str):
        name, provider = model_string.split(":")
        return cls(name, provider)

model = AIModel.from_string("GPT-5:OpenAI")
print(model.name)
print(model.provider)
```

```python
class SoftwareEngineer:
    def __init__(self, name, experience):
        self.name = name
        self.experience = experience

    def describe(self):
        return f"{self.name} has {self.experience} years of experience in software engineering."

class AIEngineer(SoftwareEngineer):

    def __init__(self, name, experience, ai_specialization):
        super().__init__(name, experience)
        self.ai_specialization = ai_specialization

    def describe(self):
        base_description = super().describe()
        return f"{base_description} Specializes in {self.ai_specialization}."

ai_eng1 = AIEngineer("Alice", 5, "Natural Language Processing")
print(ai_eng1.describe())
```

```python
import asyncio
import time

async def download_doc():
    print("downloading started")
    await asyncio.sleep(3)
    print("downloading ended")

async def generate_embeddings():
    print("generating embeddings")
    await asyncio.sleep(2)
    print("embeddings generated")

async def fetch_metadata():
    print("fetching metadata")
    await asyncio.sleep(1)
    print("metadata fetched")

async def main():
    start_time = time.perf_counter()
    await asyncio.gather(download_doc(), generate_embeddings(), fetch_metadata())
    elapsed_time = time.perf_counter() - start_time
    print(f"All tasks completed in {elapsed_time:.2f} seconds.")

await main()
```

```python
async def get_user(user_id):
    async with httpx.AsyncClient() as client:
        response = await client.get(f"https://jsonplaceholder.typicode.com/users/{user_id}")
        response.raise_for_status
        return response.json()
async def get_posts(user_id):
    async with httpx.AsyncClient() as client:
        response = await client.get(f"https://jsonplaceholder.typicode.com/posts/{user_id}")
        response.raise_for_status
        return response.json()
async def get_todos(user_id):
    async with httpx.AsyncClient() as client:
        response = await client.get(f"https://jsonplaceholder.typicode.com/todos/{user_id}")
        response.raise_for_status
        return response.json()

async def main():
    try:
        user, posts, todos = await asyncio.gather(get_user(1), get_posts(1), get_todos(1))
        return {
            "user": user,
            "posts": posts,
            "todos": todos
        }
    except Exception as ex:
        print(ex)

if __name__ == "__main__":
    res = await main()
    for k, v in res.items():
        print(k, v)
```

```python
from typing import List
from pydantic import BaseModel, Field

# 1. Define child models
class Address(BaseModel):
    street: str
    city: str
    postal_code: str
    country: str = "USA"

class Customer(BaseModel):
    id: int
    name: str
    email: str

class Product(BaseModel):
    product_id: int
    name: str
    price: float = Field(gt=0, description="Price must be strictly positive")
    quantity: int = Field(ge=1, description="Quantity must be at least 1")

# 2. Define parent model with nested fields and collections
class Order(BaseModel):
    order_id: str
    customer: Customer          # Nested Customer model
    shipping_address: Address   # Nested Address model
    items: List[Product]        # Collection of Product models

    @property
    def total_amount(self) -> float:
        """Calculates total order cost based on item prices and quantities."""
        return sum(item.price * item.quantity for item in self.items)


# 3. Example Usage & Input Validation

# Raw nested data (e.g., coming from an API request body)
order_payload = {
    "order_id": "ORD-100293",
    "customer": {
        "id": 101,
        "name": "Sarah Connor",
        "email": "sarah@example.com"
    },
    "shipping_address": {
        "street": "123 Main St",
        "city": "Los Angeles",
        "postal_code": "90001"
    },
    "items": [
        {"product_id": 1, "name": "Wireless Keyboard", "price": 49.99, "quantity": 1},
        {"product_id": 2, "name": "Ergonomic Mouse", "price": 29.50, "quantity": 2}
    ]
}

# Instantiate and validate nested structure
order = Order(**order_payload)

# 4. Accessing Nested Fields via Attribute Access
print("--- Order Summary ---")
print(f"Order ID: {order.order_id}")
print(f"Customer Name: {order.customer.name}")
print(f"Shipping City: {order.shipping_address.city}")

print("\n--- Order Items ---")
for item in order.items:
    print(f"- {item.name}: ${item.price} x {item.quantity}")

print(f"\nTotal Order Value: ${order.total_amount:.2f}")
```

```python
from typing import Dict, Any, Optional
from pydantic import BaseModel, Field, SecretStr, field_validator

class AppConfig(BaseModel):
    # 1. Sensitive API key masked with SecretStr
    api_key: SecretStr = Field(..., description="API key for external LLM provider")
    
    # 2. Environment and name fields with constraints
    app_name: str = Field(..., min_length=2, max_length=50, description="Application identifier")
    environment: str = Field(
        default="development", 
        description="Deployment stage (development, staging, production)"
    )
    
    # 3. Model configuration dictionary
    model_config_params: Dict[str, Any] = Field(
        default_factory=dict,
        description="Hyperparameters and model settings (e.g., temperature, max_tokens)"
    )
    
    # 4. Optional field
    timeout_seconds: Optional[int] = Field(default=30, ge=1, le=300)

    # Constraint to ensure environment is valid
    @field_validator("environment")
    @classmethod
    def validate_environment(cls, value: str) -> str:
        valid_envs = {"development", "staging", "production"}
        if value.lower() not in valid_envs:
            raise ValueError(f"Environment must be one of {valid_envs}")
        return value.lower()


# --- Demonstration & Verification ---

# 1. Instantiate the AppConfig model
config = AppConfig(
    api_key="sk-proj-1234567890abcdef1234567890abcdef",
    app_name="AI-Agent-Gateway",
    environment="production",
    model_config_params={
        "model_name": "gpt-4o",
        "temperature": 0.2,
        "max_tokens": 2048
    },
    timeout_seconds=60
)

print("--- Loaded Application Configuration ---")
print(f"App Name: {config.app_name}")
print(f"Environment: {config.environment}")
print(f"Model Config: {config.model_config_params}")

# Direct access to SecretStr masks raw output
print(f"Direct API Key Output: {config.api_key}")
# Accessing actual secret value
print(f"Decrypted Secret Value: {config.api_key.get_secret_value()}")


# 2. Serialize Configuration (JSON Dump)
print("\n--- Serialized JSON Output ---")
json_output = config.model_dump_json(indent=2)
print(json_output)
```

```python
import json
from typing import List, Dict, Any, Optional
from pydantic import BaseModel, Field, field_validator

# 1. Define Structured LLM Response Model
class LLMResponse(BaseModel):
    answer: str = Field(..., min_length=1, description="The main text generated by the LLM")
    confidence: float = Field(..., ge=0.0, le=1.0, description="Confidence score between 0.0 and 1.0")
    sources: List[str] = Field(default_factory=list, description="List of source URLs or references")
    metadata: Dict[str, Any] = Field(default_factory=dict, description="Execution metadata like token count or latency")

    # Custom Business Rule Validation
    @field_validator("sources")
    @classmethod
    def validate_high_confidence_sources(cls, sources: List[str], info) -> List[str]:
        """
        Custom Rule: If confidence score is provided and high (>0.8), 
        ensure at least one source citation is present.
        """
        confidence = info.data.get("confidence", 0.0)
        if confidence > 0.8 and not sources:
            raise ValueError("High confidence responses (>0.8) must cite at least one source.")
        return sources


# --- Pipeline Simulation ---

# Step 1: Raw JSON Response from external LLM API
raw_llm_json = """
{
    "answer": "Asyncio allows running concurrent I/O operations on a single thread.",
    "confidence": 0.95,
    "sources": ["https://docs.python.org/3/library/asyncio.html"],
    "metadata": {
        "model": "gpt-4o",
        "prompt_tokens": 120,
        "completion_tokens": 45
    }
}
"""

print("--- Step 1: Raw JSON Response ---")
print(raw_llm_json.strip())


# Step 2: Pydantic Validation & Instantiation (JSON -> Python Object)
llm_object = LLMResponse.model_validate_json(raw_llm_json)

print("\n--- Step 2 & 3: Validated Python Object ---")
print(f"Object Type: {type(llm_object)}")
print(f"Answer: {llm_object.answer}")
print(f"Confidence: {llm_object.confidence}")
print(f"Sources: {llm_object.sources}")


# Step 4: Application Logic
# Example: Post-processing the validated object inside the application
if llm_object.confidence >= 0.9:
    status_flag = "VERIFIED_HIGH_CONFIDENCE"
else:
    status_flag = "NEEDS_HUMAN_REVIEW"

# Append calculated application field into metadata
llm_object.metadata["status_flag"] = status_flag
llm_object.metadata["processed_by"] = "Pipeline-Worker-01"

print("\n--- Step 4: Application Logic Applied ---")
print(f"Assigned Status Flag: {status_flag}")


# Step 5 & 6: Serialization (Python Object -> JSON String)
final_json_output = llm_object.model_dump_json(indent=2)

print("\n--- Step 5 & 6: Re-Serialized Final JSON Output ---")
print(final_json_output)
```
