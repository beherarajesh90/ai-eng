```python
def calculateDiscount(raw_price: str, raw_quantity: str, raw_discount):
    price = float(raw_price)
    quantity = int(raw_quantity)
    discount = float(raw_discount)

    total_cost = price * quantity

    if(discount > 0):
        total_cost = total_cost - (total_cost * discount / 100)
    
    if(total_cost is None):
        total_cost = 0
        print("Total cost is None, setting to 0")
    
    if(not total_cost):
        total_cost = 0
        print("Total cost is 0, setting to 0")

    print(0 is None, not 0)

    return total_cost

calculateDiscount("100", "2", "10")
```

```python
def generate_llm_prompt(system_role: str, user_query: str, temperature:float):
    user_query = user_query.strip()
    return f"[SYSTEM: upper({system_role}) | TEMP: {temperature: 0.2f}] {user_query}"

generate_llm_prompt(" system assistant ", " what is rag?", 0.7)
```

```python
responses = [500, 502,503, 200, 404]

retry_after = 1
attempt = 0
for status in responses:

    if(status is None):
        print("Status code is None, skipping")
        continue

    if(status in [500,502,503]):
        print(f"Retrying after {retry_after} seconds for status code {status}")
        retry_after *= 2
        attempt +=1
    elif(status == 200):
        print("Request successful, no retry needed")
        break
    
```

```python
from datetime import datetime

def add_log_entry(
    event: str,
    level: str = "INFO",
    logs: list[str] | None = None
) -> list[str]:
    if logs is None:
        logs = []

    timestamp = datetime.now().strftime("%Y-%m-%d %H:%M:%S")
    logs.append(f"[{timestamp}] [{level}] {event}")

    return logs

logs = add_log_entry("Application started")
print(logs)
logs = add_log_entry("Database connected", "DEBUG")
print(logs)
```
